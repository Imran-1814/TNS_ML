2ans)list are mutable,ordered and uses generally changing data(School student marks) also help in retrieving data using indexing and slicing and slower and uses more memory compare to tuple
-tuples are immutable,ordered and uses generally fixed data(coordinates) also help in retrieving data using indexing and slicing and uses less memory,faster than list
-sets are mutable,unordered and uses generally unique data(School student ids) not help in retrieving data using indexing and slicing
