```python
#problem 4
st1_marks=int(input("Enter the marks of first student= "))
st2_marks=int(input("Enter the marks of second student= "))
st3_marks=int(input("Enter the marks of third student= "))
st4_marks=int(input("Enter the marks of fourth student= "))
st5_marks=int(input("Enter the marks of fifith student= "))
#calcute
total_marks=st1_marks+st2_marks+st3_marks+st4_marks+st5_marks
avg_marks=(total_marks/5)
print("Total marks of five students are",total_marks)
print("Average marks of five students are",avg_marks)
```
