```python
#To check whether a number is palindrome
num=int(input("Enter a number: "))
Original=num
rev=0
while num>0:
  digit=num%10
  rev=rev*10+digit
  num=num//10
if(Original==rev):
  print("Number is a palindrome")
else:
  print("Not a palindrome")
```

Fibonacci series is a sequence of numbers where the sum of two gives the third number


```python
#Fibonacci series
n=int(input("Enter number of terms: "))
a=0
b=1
for i in range(n):
  print(a, end=',')
  a,b=b,a+b
```

```python
#To find factorial of a number
n=int(input("Enter a number: "))
fact=1
for i in range(1,n+1):
  fact=fact*i
print("Factorial: ",fact)
```

```python

# 1. Create strings
s = "hello PYTHON world"
print("Original string:", s)

# 2. Changing case
print("capitalize:", s.capitalize())
print("casefold:", s.casefold())
print("lower:", s.lower())
print("upper:", s.upper())
print("title:", s.title())
print("swapcase:", s.swapcase())

# 3. Searching and counting
s = "python programming python"
print("count:", s.count("python"))
print("find:", s.find("programming"))
print("index:", s.index("python"))
print("rfind:", s.rfind("python"))
print("rindex:", s.rindex("python"))
print("startswith:", s.startswith("python"))
print("endswith:", s.endswith("python"))

# 4. Removing and replacing
s = "  hello python  "
print("strip:", s.strip())
print("lstrip:", s.lstrip())
print("rstrip:", s.rstrip())
print("removeprefix:", "unhappy".removeprefix("un"))
print("removesuffix:", "notes.txt".removesuffix(".txt"))
print("replace:", "hello world".replace("world", "Python"))

# 5. Splitting and joining
s = "apple,banana,mango"
print("split:", s.split(","))
print("rsplit:", s.rsplit(",", 1))
print("splitlines:", "hello\nworld".splitlines())
print("partition:", "a-b-c".partition("-"))
print("rpartition:", "a-b-c".rpartition("-"))
print("join:", "-".join(["a", "b", "c"]))

# 6. Checking string contents
print("isalnum:", "Python123".isalnum())
print("isalpha:", "Python".isalpha())
print("isascii:", "Python123".isascii())
print("isdecimal:", "123".isdecimal())
print("isdigit:", "123".isdigit())
print("isidentifier:", "my_variable".isidentifier())
print("islower:", "hello".islower())
print("isnumeric:", "123".isnumeric())
print("isprintable:", "Hello!".isprintable())
print("isspace:", "   ".isspace())
print("istitle:", "Hello World".istitle())
print("isupper:", "PYTHON".isupper())

# 7. Alignment and padding
s = "Python"
print("center:", s.center(10, "-"))
print("ljust:", s.ljust(10, "-"))
print("rjust:", s.rjust(10, "-"))
print("zfill:", "42".zfill(5))
print("expandtabs:", "a\tb".expandtabs(4))

# 8. Formatting
print("format:", "My name is {} and I am {}".format("Sam", 20))
print("format_map:", "{name} is {age}".format_map(
    {"name": "Sam", "age": 20}
))

# 9. Encoding and translation
print("encode:", "Hello".encode("utf-8"))
table = str.maketrans("ae", "12")
print("maketrans + translate:", "apple".translate(table))

# 10. Additional string method
print("encode another example:", "Python".encode())

# Check all public string methods
print("All public methods:", [
    m for m in dir(str) if not m.startswith("_")
])
```

```python
#Armstrong number

num = int(input("Enter a number: "))
temp = num
digits = len(str(num))
total = 0
while temp > 0:
    digit = temp % 10
    total += digit ** digits
    temp //= 10
if num >= 0 and total == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")

```
